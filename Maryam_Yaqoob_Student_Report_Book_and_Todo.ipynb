## Task 1: Student Report Book

```python
"""
Task 1: Student Report Book

Assumptions (flagged):
1. Each student is entered with one marks value, so that value is the
   student's average marks. The grade book also shows the class average.
2. "Restart the whole program" means all saved records are cleared and the
   menu starts again from the beginning.
3. Roll numbers are not checked for duplicates (the task does not ask for it).
"""


class RestartProgram(Exception):
    """Raised by restart_program() to send control back to a fresh start."""


def restart_program(message):
    """Show an error and restart the program."""
    print(f"\nError: {message}")
    print("Restarting the program...\n")
    raise RestartProgram


def get_grade(average):
    """Return the grade letter for an average mark."""
    if average >= 80:
        return "A"
    if average >= 70:
        return "B"
    if average >= 60:
        return "C"
    return "F"


def add_student(records):
    """Ask for name, roll number and marks, validate them, then save."""
    name = input("Enter student name: ").strip()
    if not name or not all(ch.isalpha() or ch == " " for ch in name):
        restart_program("Name must contain letters and spaces only.")

    roll = input("Enter roll number: ").strip()
    if not (roll.isascii() and roll.isdigit()):
        restart_program("Roll number must contain digits only.")

    marks = input("Enter marks (0-100): ").strip()
    if not (marks.isascii() and marks.isdigit()) or not 0 <= int(marks) <= 100:
        restart_program("Marks must be a whole number from 0 to 100.")

    records.append({"name": name, "roll": roll, "marks": int(marks)})
    print(f"Student '{name}' saved successfully.")


def view_gradebook(records):
    """Show all students, their average marks, and grade."""
    if not records:
        print("\nNo students have been added yet.")
        return

    print("\n" + "=" * 52)
    print(f"{'Name':<20}{'Roll No':<12}{'Average':<10}{'Grade':<6}")
    print("-" * 52)
    for rec in records:
        grade = get_grade(rec["marks"])
        print(f"{rec['name']:<20}{rec['roll']:<12}{rec['marks']:<10}{grade:<6}")
    print("-" * 52)
    class_avg = sum(r["marks"] for r in records) / len(records)
    print(f"Class average: {class_avg:.2f} (Grade {get_grade(class_avg)})")
    print("=" * 52)


def main():
    """Show the menu and run until the user chooses Exit."""
    records = []
    while True:
        print("\n--- Student Report Book ---")
        print("1. Add Student")
        print("2. View Grade Book")
        print("3. Exit")
        choice = input("Enter your choice (1-3): ").strip()

        if choice not in ("1", "2", "3"):
            restart_program("Menu choice must be 1, 2, or 3.")
        elif choice == "1":
            add_student(records)
        elif choice == "2":
            view_gradebook(records)
        else:
            print("Goodbye!")
            break


def run_task1():
    """Run main(); on any wrong input start again from scratch."""
    while True:
        try:
            main()
            break
        except RestartProgram:
            continue
```

Run Task 1 (type your inputs when asked). Output below is a sample run.

```python
run_task1()
```

## Task 2: Lite TO-DO List App (only two features)

```python
"""
Task 2: Lite TO-DO List App (only two features)

Features: add a task, view all tasks. A list stores the tasks.
No delete, edit, or advanced features.

Assumption (flagged): the menu also has an Exit option so the program can
stop. Exit is not a third feature, just a way to leave the loop.
"""


def add_task(tasks):
    """Let the user type a new task and store it in the list."""
    task = input("Enter a new task: ").strip()
    if task:
        tasks.append(task)
        print("Task added.")
    else:
        print("Task cannot be empty.")


def view_tasks(tasks):
    """Show all tasks."""
    if not tasks:
        print("\nYour to-do list is empty.")
        return
    print("\nYour tasks:")
    for number, task in enumerate(tasks, start=1):
        print(f"{number}. {task}")


def run_task2():
    tasks = []  # a list stores the tasks
    while True:
        print("\n--- Lite TO-DO List ---")
        print("1. Add a task")
        print("2. View all tasks")
        print("3. Exit")
        choice = input("Enter your choice (1-3): ").strip()

        if choice == "1":
            add_task(tasks)
        elif choice == "2":
            view_tasks(tasks)
        elif choice == "3":
            print("Goodbye!")
            break
        else:
            print("Invalid choice. Please enter 1, 2, or 3.")
```

Run Task 2 (type your inputs when asked). Output below is a sample run.

```python
run_task2()
```
